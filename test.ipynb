```python
import matplotlib.pyplot as plt, numpy as np, matplotx, pandas as pd
plt.style.use(matplotx.styles.ayu['light'])
```

```python
a = -0.5
b = 5
f = lambda x: x ** 2 - 3

df = pd.DataFrame(columns= ['a', 'b', 'c', 'f(a)', 'f(b)', 'f(c)', 'itr'])
df2 = pd.DataFrame(columns= ['err', 'itr'])

for i in range(10):
    c = (a + b) / 2
    df.loc[len(df)] = {
        'a' : a,
        'b' : b,
        'c' : c,
        'f(a)' : f(a),
        'f(b)' : f(b),
        'f(c)' : f(c),
        'itr' : i + 1
    }

    df2.loc[len(df2)] = {
        'err' : (c - 3 ** 0.5) ** 2,
        'itr' : i + 1
    }

    if f(a) * f(c) < 0:
        a, b = a, c
    elif f(b) * f(c) < 0:
        a, b = b, c

df2
```

```python

x = np.linspace(a, b, 1000)
plt.axhline(y = 0, linestyle = "dashed", alpha = 0.6)
plt.axvline(x = a, linestyle = "dashed", alpha = 0.45, color = 'red')
plt.axvline(x = b, linestyle = "dashed", alpha = 0.45, color = 'red')
plt.axvline(x = 0, linestyle = "dashed", alpha = 0.6)
plt.plot(x, f(x))
```
