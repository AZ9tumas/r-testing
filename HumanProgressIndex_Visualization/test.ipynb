```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv('data.csv', encoding='latin-1')
```

```python
print(*df.columns, sep='\n')
```

```python
# key columns: country, region, hdicode, hdi_rank_2022
# year 1990 to 2022
df[['country', 'region', 'hdicode', 'hdi_rank_2022'] + ["hdi_" + str(year) for year in range(1990, 2023)]].head()

newdf = pd.DataFrame()

for year in range(1990, 2023):
    # new structure: country, region, hdicode, hdi_rank_2022, hdi, le, eys, gni, pop, year
    temp = df[['country', 'region', 'hdicode', 'hdi_rank_2022']].copy()
    temp['hdi'] = df[f'hdi_{year}']
    temp['le'] = df[f'le_{year}']
    temp['eys'] = df[f'eys_{year}']
    temp['gni'] = df[f'gnipc_{year}']
    temp['pop'] = df[f'pop_total_{year}']
    temp['year'] = year
    newdf = pd.concat([newdf, temp], ignore_index=True)

newdf.head()
```

```python
# save df as csv with name "final_data.csv"
newdf.to_csv("final_data.csv", index=False)
```
